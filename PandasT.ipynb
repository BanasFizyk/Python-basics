```python
def gaussian(x, mu, sigma):
    return (1/(sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

x_values = np.linspace(-10, 10, 100)

plt.figure(figsize=(10, 6))
plt.plot(x_values, gaussian(x_values, mu=0, sigma=1), label='Gaussian (mu=0, sigma=1)')
plt.plot(x_values, gaussian(x_values, mu=2, sigma=0.5),
            label='Gaussian (mu=2, sigma=0.5)')
plt.plot(x_values, gaussian(x_values, mu=-2, sigma=1.5),
            label='Gaussian (mu=-2, sigma=1.5)')
plt.title('Gaussian Functions')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid()
plt.show()
```
