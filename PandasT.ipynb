```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sympy as sp
```

```python
x1 = sp.Symbol('x1')
x2 = sp.Symbol('x2')
x3 = sp.Symbol('x3')

X = np.array([x1, x2, x3])
Y = np.array([[2], [3], [4]])

Z = X*Y

print(Z)
```

```python
df = pd.read_excel('Dane_przyklad.xlsx', sheet_name='Arkusz1') #wczytanie excela (arkusz1)
print(df)

V = np.array(df.loc[1:, ["Unnamed: 1"]]) # lokalizuje kolumne V [mV] (Unnamed: 1) pomijajac naglowek (1:)
E = np.array(df.loc[1:, ["Unnamed: 2"]]) # lokalizuje kolumne E [keV] (Unnamed: 2) pomijajac naglowek (1:)

plt.scatter(V, E)
plt.xlabel('V [mV]')
plt.ylabel('E [keV]')
plt.grid(True)
plt.show()
```
