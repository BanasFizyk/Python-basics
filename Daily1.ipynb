$ \textbf{ZADANIE 1} $

Program pobiera liczbę całkowitą i wypisuje:

    -czy jest dodatnia, ujemna, czy równa zero;
    -czy jest parzysta, czy nieparzysta;
    -czy jest podzielna jednocześnie przez 3 i 5.

```python
#zadanie 1

x1 = int(input("Podaj liczbę: "))

def znak(x):
    if x>0:
        print("liczba jest dodatnia")
    elif x<0:
        print("liczba jest ujemna")
    else:
        print("liczba jest równa 0")

def reszt(x):
    if x%2==0:
        print("liczba jest parzysta")
    else:
        print("liczba jest nieparzysta")

def dziel(x):
    if x%3==0 and x%5==0:
        print("liczba jest podzielna równocześnie przez 3 i 5")
    else:
        print("liczba nie jest podzielna równocześnie przez 3 i 5")

znak(x1) # jako x1 wprowadzam 15
reszt(x1)
dziel(x1)
```

$ \textbf{ZADANIE 2} $

Mając listę

$ \text{measurements} = [12.5, -1.0, 14.8, 17.2, -3.5, 11.9, 20.1] $

Wartości ujemne oznaczają błędne pomiary. Napisz program, który:

    -utworzy nową listę zawierającą tylko poprawne pomiary;
    -policzy ich sumę i średnią;
    -znajdzie wartość najmniejszą oraz największą;
    -policzy, ile poprawnych pomiarów przekracza 15.

Nie usuwaj elementów z pierwotnej listy podczas przechodzenia po niej.

```python
#zadanie 2

measurements = [12.5, -1.0, 14.8, 17.2, -3.5, 11.9, 20.1]


m_x = measurements.copy()
x = int(len(measurements))

X=[]

for i in range(x):
    p = m_x[i]
    if p<0:
        X.append(p)

for j in X:
    m_x.remove(j)

print(measurements)
print(m_x)

m_xS = sum(m_x)
print(f'Suma poprawnych wartości jest równa: ', m_xS)
m_xSr = m_xS/int(len(m_x))
print(f'Średnia poprawnych wartości jest równa: ', m_xSr)

m_sort = sorted(m_x)
m_min = m_sort[0]
m_max = m_sort[-1]
print(f'Wartość minimalna: ', m_min)
print(f'Wartość maksymalna: ', m_max)

y = int(len(m_x))

T=[]

for k in range(y):
    r = m_x[k]
    if r>15:
        T.append(r)

pp = int(len(T))
print(f'Liczba poprawnych pomiarów większych od 15 jest równa: ', pp)
```

$\textbf{ZADANIE 3}$

Program prosi użytkownika o liczby tak długo, aż poda 0. Następnie wypisuje:

    -ile liczb podano, bez końcowego zera;
    -ich sumę;
    -największą podaną liczbę.

Postaraj się obsłużyć również sytuację, w której pierwszą podaną liczbą jest 0.

```python
#zadanie 3 - z czata (nie umiałeś frajerze)

#while x!=0: # nie mam pojęcia jak sobie poradzić z tym zadaniem
   # x = int(input("Podaj liczbę: "))

counter = 0
total = 0
largest = None

while True:
    n = int(input("Podaj liczbę: "))

    if n == 0:
        break

    counter += 1
    total += n

    if largest is None or n > largest:
        largest = n

print(f"Liczba podanych wartości: {counter}")
print(f"Suma: {total}")

if largest is None:
    print("Nie podano żadnej liczby poza zerem.")
else:
    print(f"Największa liczba: {largest}")
```

$ \textbf{Łatwe 1 — suma liczb dodatnich} $

Pobieraj od użytkownika liczby całkowite. Program ma działać do momentu podania liczby ujemnej.

Następnie wypisz:

    -sumę podanych liczb dodatnich i zer,
    -liczbę wprowadzonych wartości — bez końcowej liczby ujemnej.

```python
suma = 0
ilosc = 0

while True:
    x = int(input("Liczba: "))

    if x<0:
        break

    if x>=0:
        ilosc += 1
        suma += x

print(f'Ilość podanych liczb oprócz ujemnej wynosi: {ilosc}')
print(f'Suma podanych liczb wynosi: {suma}')
```

$ \textbf{Łatwe 2 — zliczanie parzystych i nieparzystych} $

Pobieraj liczby aż użytkownik poda 0. Końcowego zera nie uwzględniaj.

Program ma wypisać:

    -liczbę wartości parzystych,
    -liczbę wartości nieparzystych.

```python
n_even = 0
n_odd = 0

while True:
    x = int(input("Podaj liczbę: "))

    if x == 0:
        break

    if x%2 == 0:
        n_even += 1

    if x%2 != 0:
        n_odd += 1

print(f'Ilość liczb parzystych: {n_even}')
print(f'Ilość liczb nieparzystych: {n_odd}')
```

```python
#Ruch jednostajnie przyspieszony

a = 3 #m/s^2
v_0 = 2 #m/s

velocities = [] #m/s
positions = [] #m

times = [0, 1, 2, 3, 4, 5] #s

# alternatywa 
V = [a*t + v_0 for t in times]

for t in times:
    v = a*t + v_0
    x = 0.5*a*(t**2) + v_0*t
    velocities.append(v)
    positions.append(x)

print(f"{'Czas [s]':<14} | {'Prędkość [m/s]':<14} | {'Położenie [m]':<14}")
print("-" *48)

for x, y, z in zip(times, velocities, positions): #zip łączy listy w takie fajne paczki
    print(f"{x:<14} | {y:<14} | {z:<14} ")

print(" ")
print(f'Największa prędkość: {max(V)} m/s')
print(f'Położenie końcowe: {positions[-1]} m')

for i in range(len(V)):
    if V[i] <= 10:
        continue
    else:
        print(f'Wartość chwili dla której prędkość przekroczyła 10 m/s wynosi: {times[i]}')
        break
```

```python
#analiza energii kinetycznej

velocities = []

while True:
    m = float(input("Podaj masę ciała w [kg]: "))

    if m > 0:
        break
    else:
        print("Niepoprawna wartość!")
        continue

n_con = 0

while True:
    v = float(input("Podaj wartości prędkości w [m/s]: "))

    if v == -1:
        break

    if v < 0 and v != -1: #nie potrzebny drugi warunek bo został już sprawdzony
        print("Niepoprawna wartość!")
        continue
    else:
        velocities.append(v)

    n_con +=1

energies = [0.5*m*(v**2) for v in velocities]

if n_con > 0:
    print(f"Masa ciała wynosi: {m}kg")
    print(" ")
    #print(*velocities) #poprawne i spoko ale brzydkie 
    #print(*energies)

    #velocities_text = ", ".join(map(str, velocities))
    #print(f"{'Prędkości [m/s]:':<36} | {velocities_text}") #lepiej bo zagnieżdżone "" może dać błąd

    print(f"{'Prędkości [m/s]: ':<36} | {", ".join(map(str, velocities)):<8}")
    print(f"{'Energie kinetyczne [J]: ':<36} | {", ".join(map(str, energies)):<8}")
    print(" ")
    print(f'Liczba pomiarów: {int(len(energies))}')
    print(f'Największa wartość energii: {max(energies)}J')
    print(f'Najmniejsza wartość energii: {min(energies)}J')

    num = 0

    for k in energies:
        if k > 100:
            num += 1

    print(f'Ilość pomiarów w których energia była większa niż 100J: {num}')

else:
    print("Nie podano żadnej warości!")
```

```python
#Zasięg rzutu ukośnego

v_0 = 20 #m/s
g = 9.81 #m/s^2

angles = [5, 10, 15, 20, 25, 30, 35, 40, 45,
          50, 55, 60, 65, 70, 75, 80, 85]
angles_rad = []

ranges = []
ranges_ort = []

for a in angles:
    alpha = math.radians(a)
    angles_rad.append(alpha)

    R = (v_0**2 * math.sin(2*alpha) ) / g
    ranges.append(R)
    R_ort = (v_0**2 * math.sin(math.pi-2*alpha) ) / g
    ranges_ort.append(R_ort)


print(f"{'Miara kąta [°]':<18} | {'Miara kąta [rad]':<18} | {'Zasięg rzutu [m]':<18}")
print("-" *58)
for x,y,z in zip(angles, angles_rad, ranges):
    print(f"{x:<18} | {round(y,4):<18} | {round(z,4):<18}")

print(" ")

angles_l = []

r_max = max(ranges) # aby nie liczyć cały czas w pętli

for i in range(int(len(ranges))):
    if ranges[i] > 30:
        r = angles[i]
        angles_l.append(r)
    if ranges[i] == r_max:
        print(f"Największy zasięg wynosi: {round(ranges[i],4)}m, dla kąta: {angles[i]}°")
        print(" ")

print("Kąty w dla których zasięg był większy niż 30m: ")
print("\n", "\t", *angles_l, "[°]")

# Sprawdzenie czy wartości są sobie równe:

#x = math.isclose(ranges[1], ranges_ort[1])
#print(x)

print(" ")

for i in range(int(len(ranges))):
    x = math.isclose(ranges[i], ranges_ort[i])
    if x == False:
        print("Wartości nie są takie same!")
        break
else:
    print("Wartości zasięgu dla 2*alpha i 180°-2*alpha są takie same!")
```
