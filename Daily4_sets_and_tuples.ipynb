```python
numbers = [4.5, 8.2, 1.3, 9.7, 6.4]

def calculate_statistics(numbers):
    if not numbers:
        return None
    else:
        minimum = min(numbers)
        maximum = max(numbers)
        average = sum(numbers)/len(numbers)

        return minimum, maximum, average

if not numbers:
    print("lista jest pusta")
else:
    mini, maxi, avg = calculate_statistics(numbers)
    print(mini)
    print(maxi)
    print(avg)
```

$ \textbf{Zadanie 1 — analiza zdarzeń detektora} $

Dane zapisano jako listę krotek. Napisz funkcje która:

    -pomija niepoprawne zdarzenia;
    -tworzy zbiór wszystkich wykrytych rodzajów cząstek;
    -tworzy słownik liczników;
    -tworzy zbiór cząstek, których przynajmniej jedno poprawne zdarzenie miało energię powyżej 15;
    -znajduje poprawne zdarzenie o największej energii;
    -zwraca krotkę: (unique_particles, counters, high_energy_particles, strongest_event)

```python
events = [
    (1001, "electron", 12.5, True),
    (1002, "proton", 8.2, True),
    (1003, "electron", 15.1, False),
    (1004, "photon", 22.7, True),
    (1005, "proton", 9.4, True),
    (1006, "muon", 18.3, True),
    (1007, "photon", 21.5, False),
    (1008, "electron", 14.8, True)
]

counters = {}

def analyze_events(events):
    names = []
    val_15 = set()
    n_max = None
    name_max = None
 
    for item in events:
        p_id, name, value, valid = item
        if valid == True:
            names.append(name)
            if value > 15:
                val_15.add(name)
            if n_max == None or value > n_max:
                n_max = value
                name_max = name

    names_set = set()
    for i in names:
        if i not in names_set:
            counters[i] = 1
            names_set.add(i)
        else:
            counters[i] += 1
   
    return names_set, val_15, name_max, n_max, counters

names, val_15, name_max, n_max, counters = analyze_events(events)

print(names)
print(val_15)
print(f"{name_max}: {n_max}")
print(counters.items())
```

$ \textbf{Zadanie 2 — kontrola dostępu} $

Zasady:

    -dostęp jest przyznany, jeżeli użytkownik ma wszystkie wymagane uprawnienia;
    -dostęp jest odrzucony, jeśli brakuje choć jednego wymaganego uprawnienia;
    -dostęp jest również odrzucony, jeśli użytkownik ma dowolne zabronione uprawnienie.

```python
required_permissions = {"read", "write"}
forbidden_permissions = {"delete_database", "root"}

users = {
    "Anna": {"read", "write", "execute"},
    "Jan": {"read"},
    "Ola": {"read", "write", "root"},
    "Piotr": {"read", "write"}
}


def check_access(user_permissions, required, forbidden):
    if required.issubset(user_permissions) == True and forbidden.isdisjoint(user_permissions) == True:
        return True, set(), set()
    if required.issubset(user_permissions) == True and forbidden.isdisjoint(user_permissions) == False:
        idj = forbidden & user_permissions
        return False, set(), idj
    if required.issubset(user_permissions) == False and forbidden.isdisjoint(user_permissions) == True:
        nrq = required - user_permissions
        return False, nrq, set()
    if required.issubset(user_permissions) == False and forbidden.isdisjoint(user_permissions) == False:
        nrq = required - user_permissions
        idj = forbidden & user_permissions
        return False, nrq, idj
        

for name, user_permissions in users.items():
    access, missing, forbidden = check_access(user_permissions, required_permissions, forbidden_permissions)
    if access == True:
        print(f"Dostęp dla {name} przyznany.")
        print(" ")
    else:
        if missing == set():
            print(f"Dostęp dla {name} odrzucony. \nUżytkownik posiada zabronione uprawnienie: {forbidden}")
            print(" ")
        else:
            print(f"Dostęp dla {name} odrzucony. \nUżytkownik nie posiada uprawnienia: {missing}")
            print(" ")
```

$ \textbf{Zadanie 3 — mapa trafień detektora} $

Zadanie 3 — mapa trafień detektora. Napisz funkcję, która tworzy słownik. Następnie napisz funkcję która bez używania max(..., key=...) zwraca: $ \textit{position, number of hits} $. Na końcu:

    -wypisz liczbę wszystkich trafień;
    -wypisz liczbę różnych aktywnych komórek;
    -wypisz najbardziej aktywną komórkę;
    -utwórz zbiór komórek, które zarejestrowały co najmniej dwa trafienia.

```python
hits = [
    (0, 0),
    (1, 2),
    (0, 0),
    (2, 1),
    (1, 2),
    (1, 2),
    (3, 0),
    (2, 1),
    (0, 3)
]

n_hits = len(hits)
print(n_hits)


def count_hits(hits):
    hit_counter = {}
    hits_set = set()
    for i in hits:
        if i not in hits_set:
            hits_set.add(i)
            hit_counter[i] = 1
        else:
            hit_counter[i] += 1

    dif_hit = len(hits_set)

    return hit_counter, dif_hit

hit_counter, dif_hit = count_hits(hits)

print(dif_hit)
print(hit_counter)

def find_most_active_cell(hit_counter):
    position = None
    hit_max = None
    for key, value in hit_counter.items():
        if hit_max == None or value > hit_max:
            hit_max = value
            position = key

    return position, hit_max

x, y = find_most_active_cell(hit_counter)

print(f"{x}: {y}")

set_two = {
            key
            for key, value in hit_counter.items()
            if value >= 2
}

print(set_two)
```

$ \textbf{Zadanie 4 — porównanie dwóch eksperymentów} $

```python
experiment_a = [
    ("electron", 12.5),
    ("proton", 8.4),
    ("photon", 20.1),
    ("muon", 17.2)
]

experiment_b = [
    ("electron", 13.1),
    ("neutron", 9.8),
    ("photon", 21.4),
    ("tau", 25.0)
]

def extract_particles(experiment):
    particles_names = set()
    for item in experiment:
        name, value = item
        particles_names.add(name)

    return particles_names

particles_a = extract_particles(experiment_a)
particles_b = extract_particles(experiment_b)

print(particles_a & particles_b)
print(particles_a - particles_b)
print(particles_b - particles_a)
print(particles_a ^ particles_b)
print(particles_a | particles_b)

all_events = experiment_a + experiment_b

maximum_energies = {}

for name, energy in all_events:
    if name not in maximum_energies:
        maximum_energies[name] = energy
    elif energy > maximum_energies[name]:
        maximum_energies[name] = energy

print(maximum_energies)
```
