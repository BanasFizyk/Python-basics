```python
n = 0

with open("particles.txt", "r", encoding = "UTF-8") as particles_file:
    for line in particles_file:
        print(line.strip())
        n += 1

    print(n)
```

```python
measurements_dict = {}

try:
    with open("measurements.txt", "r") as measurements:
        for line_number, line in enumerate(measurements, start=1):
            try:
                name, value_txt = line.split(",")
                value = float(value_txt)

                if value <= 0:
                    raise ValueError(f"Energy cannot be negative or 0!")
                if not name:
                    raise ValueError(f"Particle name cannot be empty")

                measurements_dict.setdefault(name, []).append(value)

            except ValueError as error:
                print(f"In line {line_number}. {line} error: {error}")
except FileNotFoundError:
    print("The measurements file was not found.")
else:
    for key, value in measurements_dict.items():
            avg = sum(value)/len(value)
            print(f"{key} average: {avg}")
```

```python
measurements = {
    "electron": [12.5, 15.7, 14.2],
    "photon": [20.1, 18.6],
    "proton": [8.4, 9.1, 8.8]
}
try:
    with open("statistics.txt", "w", encoding="utf-8") as statistics:
        for key, value in measurements.items():
            avg = sum(value)/len(value)
            statistics.write(f"Particle: {key}\n")
            statistics.write(f"Measurements: {len(value)}\n")
            statistics.write(f"Average: {avg:.2f}\n")
            statistics.write(f"Minimum: {min(value)}\n")
            statistics.write(f"Maximum: {max(value)}\n")
            statistics.write(f"\n")
except OSError as error:
    print("File error:", error)
else:
    print("Statistics saved successfully.")
```

```python
with open("raw_data.txt", "r", encoding="utf-8") as source, \
     open("clean_data.txt", "w", encoding="utf-8") as clean_file, \
     open("errors.txt", "w", encoding="utf-8") as error_file:

    for line_n, line in enumerate(source, start=1):
        line = line.strip()

        try:
            name, value_txt = line.split(",")
            name = name.strip().lower()
            value = float(value_txt.strip())

            if value <= 0:
                raise ValueError("Energy must be positive!")
            if not name:
                raise ValueError("Particle name cannot be empty")

        except ValueError as error:
            error_file.write(f"In line {line_n}: {line} | {error}\n")
        else:
            clean_file.write(f"{name}: {value} \n")
print("Correct records: ")
with open("clean_data.txt", "r", encoding="utf-8") as clean_file:
    for line in clean_file:
        print(line.strip())

print(" ")

print("Invalid records: ")
with open("errors.txt", "r", encoding="utf-8") as error_file:
    for line in error_file:
        print(line.strip())

#zadanie 3
n_1 = 0
n_2 = 0
with open("clean_data.txt", "r", encoding="utf-8") as clean_data, \
     open("errors.txt", "r", encoding="utf-8") as error_file, \
     open("analysis.log", "a", encoding="utf-8") as log_file:
    for line1 in clean_data:
        n_1 += 1
    for line2 in error_file:
        n_2 += 1

    log_file.write(f"Analysis completed | correct: {n_1} | invalid: {n_2} \n")
```
