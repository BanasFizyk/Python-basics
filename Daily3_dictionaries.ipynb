```python
particle = {
    "name": "electron",
    "mass": 9.109e-31,
    "charge": -1.602e-19,
    "velocity": 2.0e6
}

def kinetic_energy(particle):
    v = particle["velocity"]
    m = particle["mass"]
    return (0.5*m*(v**2))/(abs(particle["charge"]))

E = kinetic_energy(particle)
print(f'Wartość energii kinetycznej wynosi: {round(E,4)} eV')

particle["kinetic_energy"] = round(E,4)

#print(particle.items())
print(" ")

for key, value in particle.items():
    print(f"{key}: {value}")
```

$ \textbf{Zadanie 1 — uzupełnianie pomiaru} $

```python
measurement = {
    "time": 4.0,
    "initial_velocity": 3.0,
    "acceleration": 2.0
}

def complete_measurement(measurement):
    t = measurement["time"]
    v_0 = measurement["initial_velocity"]
    a = measurement["acceleration"]

    v = a*t + v_0
    x = 0.5*a*(t**2) + v_0*t

    measurement.update({
        "final_velocity": v,
        "position": x
    })

    return measurement

mea = complete_measurement(measurement)

for key, value in mea.items():
    print(f"{key}: {value}")
```

```python
experiment = {
    "mass": 5.0,
    "velocity": 12.0
}

def calculate_kinetic_energy(experiment):
    m = experiment.get("mass", None)
    v = experiment.get("velocity", None)

    if m == None or v == None or v < 0 or m < 0:
        return None
    if m > 0:
        return 0.5*m*(v**2)

E = calculate_kinetic_energy(experiment)

if E == None:
    print("Brak danych")
else:
    experiment["Kinetic_energy"] = E
    for key, value in experiment.items():
        print(f"{key}: {value}")
```

$ \textbf{Zadanie 3 — analiza ruchu planet} $

```python
objects = {
    "Earth": {
        "gravity": 9.81,
        "fall_times": [1.0, 2.0, 3.0]
    },
    "Moon": {
        "gravity": 1.62,
        "fall_times": [1.0, 2.0, 3.0]
    },
    "Mars": {
        "gravity": 3.71,
        "fall_times": [1.0, 2.0, 3.0]
    }
}

def calculate_fall_distances(objects):
    for i in objects:
        g = objects[i]["gravity"]
        t = objects[i]["fall_times"]

        h = [0.5*g*(j**2) for j in t]
        objects[i]["fall_distances"] = h

    return objects

objects_update = calculate_fall_distances(objects)

for name, data in objects_update.items():
    print(f"Nazwa obiektu: {name}")
    for key, value in data.items():
        print(f"{key}: {value}")
    print(" ")
```

```python
warehouse = {
    "screws": {
        "price": 0.15,
        "quantity": 120
    },
    "nuts": {
        "price": 0.12,
        "quantity": 80
    },
    "washers": {
        "price": 0.08,
        "quantity": 200
    }
}

# def add_product(warehouse, name, price, quantity): #to źle bo zmienia wszystko
#     if price < 0 or quantity < 0:
#         print("Podano złą wartość")
#         return None
#     else:
#         if name in warehouse:
#             for key, value in warehouse.items():
#                 value["price"] = price
#                 value["quantity"] += quantity
#         else:
#             warehouse[name] = {
#                 "price": price,
#                 "quantity": quantity
#             }

#         return warehouse

def add_product(warehouse, name, price, quantity):
    if price < 0 or quantity < 0:
        return None

    if name in warehouse:
        warehouse[name]["price"] = price
        warehouse[name]["quantity"] += quantity
    else:
        warehouse[name] = {
            "price": price,
            "quantity": quantity
        }

    return warehouse

# def remove_product(warehouse, name, quantity):
#     if name in warehouse:
#         if quantity < 0:
#             return None
#         else:
#             for key, value in warehouse.items():
#                 if value["quantity"] < quantity:
#                     return None
#                 elif value["quantity"] == quantity:
#                     value["quantity"] = 0
#                 else:
#                     value["quantity"] -= quantity

#         return warehouse
    
#     else:
#         return None

def remove_product(warehouse, name, quantity):
    if name not in warehouse:
        return None

    if quantity <= 0:
        return None

    if warehouse[name]["quantity"] < quantity:
        return None

    warehouse[name]["quantity"] -= quantity

    return warehouse

def calculate_warehouse_value(warehouse):
    total_val = 0
    for key, value in warehouse.items():
        val = value.get("price", 0) * value.get("quantity", 0)
        value["value"] = val
        total_val += val

    return warehouse, total_val

warehouse, total_val = calculate_warehouse_value(warehouse)

print(total_val)
print(f"{'name':<9} | {'price':<9} | {'quantity':<9} | {'value':<9}")
print("-" *42)
for key, value in warehouse.items():
    print(f"{key:<9} | {value["price"]:<9} | {value["quantity"]:<9} | {value["value"]:<9}")
```

$ \textbf{Zadanie 6 -- katalog ciał niebieskich} $

Dodaj do słownika grawitację, pierwszą oraz drugą prędkość kosmiczną. Następnie stwórz funkcje która pobiera słownik oraz nazwę jeśli nazwa znajduje się w słowniku zwraca słownik z wartościami dla danej planety, a jeśli nie to zwraza None. Dodatkowo stwórz compression dictionary, który przechowuje wszystkie planety których prędkość ucieczki jest większa niż $5000 \frac{\mathrm{m}}{\mathrm{s}}$.

```python
import math

objects = {
    "Earth": {
        "mass": 5.972e24,
        "radius": 6.371e6
    },
    "Moon": {
        "mass": 7.342e22,
        "radius": 1.737e6
    },
    "Mars": {
        "mass": 6.417e23,
        "radius": 3.390e6
    }
}

G = 6.67430e-11

def calculate_object_parameters(objects):
    for key, value in objects.items():
        m = value["mass"]
        r = value["radius"]

        g = (G*m)/r**2
        v_o = math.sqrt((G*m)/r)
        v_e = math.sqrt((2*G*m)/r)
        value["gravity"] = round(g, 3)
        value["orbital_velocity"] = round(v_o, 3)
        value["escape_velocity"] = round(v_e, 3)

    return objects

objects = calculate_object_parameters(objects)

planets_high_v_e = {key: value["escape_velocity"] for key, value in objects.items() if value["escape_velocity"] > 5000}

print(planets_high_v_e)

def find_object(objects, name):
    if name in objects:
        return objects[name]
    else:
        return None

name = input("Podaj nazwę planety: ")

find_object(objects, name)
```
