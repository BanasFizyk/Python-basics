```python
# electron,12.5
# photon,20.1
# electron,15.7
# proton,8.4
# ,10
# proton,empty
# photon,18.6
# photon,-10


new_lines = []

p_name = input("Particles name: ")

file_path = Path("measurements.txt")

try:
    with open(file_path, "r", encoding = "utf-8") as file:
        n_records = 0
        for line in file:
            try:
                name, value_txt = line.strip().split(",")
                value = float(value_txt)

                if not name:
                    raise ValueError("Name is empty!")
                if value <= 0:
                    raise ValueError("Energy is negative or 0!")

                if name == p_name:
                    x = float(input("New energy value: "))
                    value = x       
                    n_records += 1
                    new_lines.append(f"{name},{value}\n")
                else:
                    new_lines.append(line)

            except ValueError:
                pass

    
    with open(file_path, "w", encoding="utf-8") as file:
        file.writelines(new_lines)

except FileNotFoundError:
    print("Nieznaleziono pliku!")

print(f"Zmieniono {n_records} rekordy.")
```

```python
from pathlib import Path

file_path = Path("measurements.txt")
new_lines = []
n_records = 0

try:
    particle_name = input("Particle name: ").strip().lower()
    new_energy = float(input("New energy value: ").strip())

    if not particle_name:
        raise ValueError("Particle name cannot be empty.")

    if new_energy <= 0:
        raise ValueError("Energy must be positive.")

    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            try:
                name, value_text = line.strip().split(",")

                name = name.strip().lower()
                old_energy = float(value_text.strip())

                if not name:
                    raise ValueError("Particle name is empty.")

                if old_energy <= 0:
                    raise ValueError(
                        "Existing energy is not positive."
                    )

            except ValueError:
                # Zachowujemy wadliwą linię bez zmian.
                new_lines.append(line)

            else:
                if name == particle_name:
                    new_lines.append(
                        f"{name},{new_energy}\n"
                    )
                    n_records += 1
                else:
                    new_lines.append(line)

except FileNotFoundError:
    print("Nie znaleziono pliku.")

except ValueError as error:
    print("Invalid input:", error)

except OSError as error:
    print("File error:", error)

else:
    if n_records == 0:
        print("Particle not found. File was not changed.")

    else:
        try:
            with open(file_path, "w", encoding="utf-8") as file:
                file.writelines(new_lines)

        except OSError as error:
            print("Could not save the file:", error)

        else:
            print(f"Changed records: {n_records}")
```
