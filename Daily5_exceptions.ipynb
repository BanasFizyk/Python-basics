$ \textbf{Zadanie — rejestr pomiarów energii} $

```python
particles_energies = {}

while True:
    try:
        user_input = input("Podaj nazwę cząstki oraz jej energię: ")
        if user_input == "stop":
            break
        name, energy_text = user_input.split(",")
        energy = float(energy_text.strip())

        if energy <= 0:
            raise ValueError("Energy must be positive.")
        if not name:
            raise ValueError("Enter a particle name.")

        particles_energies.setdefault(name.strip().lower(), []).append(energy)

    except ValueError as error:
        print("Invalid measurement:", error)

print(particles_energies)

particles_statistics = {}
n = 0

for key, value in particles_energies.items():
    n += len(value)
    p_avg = sum(value)/len(value)
    p_max = max(value)
    c = len(value)

    # particles_statistics.setdefault(key, {})["average"] = p_avg
    # particles_statistics.setdefault(key, {})["maximum"] = p_max

    particles_statistics[key] = {
        "average": p_avg,
        "maximum": p_max,
        "count": c
    }

print(n)
print(particles_statistics)
```

Zadanie 1 — kalkulator laboratoryjny

Napisz program działający w pętli. Użytkownik podaje działanie w formacie:

    -liczba operator liczba

```python
def lab_calc():
    while True:
        operation = input("Podaj działanie: ")
        if operation == "stop":
            break

        try:
            x_text, operator, y_text = operation.split()
            x = float(x_text.strip())
            y = float(y_text.strip())

            if operator == "/":
                value = x / y
            elif operator == "*":
                value = x * y
            elif operator == "+":
                value = x + y
            elif operator == "-":
                value = x - y
            elif operator == "^":
                value = x ** y
            else:
                raise ValueError("Wrong operator!!!")

        except ValueError as error:
            print("Error:", error)
        except ZeroDivisionError:
            print("You can't devide by 0")
        else:
            print(value)
        finally:
            print("Calculation attempt completed.")

lab_calc()
```

$ \textbf{Zadanie 2 — analiza wielu pomiarów} $

Program powinien osobno rozpoznawać:

    -nieprawidłową liczbę elementów,
    -pustą nazwę cząstki,
    -energię niebędącą liczbą,
    -energię równą zero lub ujemną.

Wskazówka: ponieważ różne problemy mogą prowadzić do ValueError, możesz samodzielnie zgłaszać

```python
raw_measurements = [
    "electron,12.5",
    "photon,20.1",
    "muon,error",
    "proton,-4.2",
    "electron,15.7",
    "invalid measurement",
    "photon,18.6,extra",
    ",10.5",
    "neutron,9.8"
]

correct_measurements = {}
errors = []

for item in raw_measurements:
    try:
        name, value_text = item.split(",")
        value = float(value_text)

        name = name.strip().lower()

        if not name:
            raise ValueError("Particle name cannot be empty!")
        if value <= 0:
            raise ValueError("Particle energy cannot be negative or 0!")

        correct_measurements.setdefault(name, []).append(value)

    except ValueError as error:
        err_dict = {}
        err_dict["data"] = item
        err_dict["reason"] = str(error)
        errors.append(err_dict)
        

print(correct_measurements)
print(errors)

n_c = 0
n_e = len(errors)

for key, value in correct_measurements.items():
    n_c += len(value)
    avg = sum(value) / len(value)
    print(f"{key} average energy: {avg}")

print(" ")
print(f"Number of correct measurements: {n_c}")
print(f"Number of errors: {n_e}")
```

$ \textbf{Zadanie 3 — własne wyjątki} $

Możemy tworzyć własne rodzaje wyjątków:

```python
import math

class InvalidEnergyError(Exception):
    pass

class InvalidMassError(Exception):
    pass

def calculate_velocity(energy, mass):
    if mass <= 0:
        raise InvalidMassError("Mass cannot be negative or 0!")
    if energy <= 0:
        raise InvalidEnergyError("Energy cannot be negative or 0!")
    return math.sqrt((2*energy)/mass)

try:
    mass = float(input("Mass: "))
    energy = float(input("Energy: "))

except InvalidMassError as error:
    print("Error: ", error)
except InvalidEnergyError as error:
    print("Error: ", error)
except ValueError as error:
    print("Error: ", error)
else:
    velocity = calculate_velocity(energy, mass)
    print(f"Velocity of object: {velocity}")
```
